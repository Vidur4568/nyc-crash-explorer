```python
cleaned = df.copy()

cleaned['CRASH DATE'] = pd.to_datetime(cleaned["CRASH DATE"])

cleaned["BOROUGH"] = cleaned["BOROUGH"].fillna("Unknown")
```

```python
cleaned.isna().sum()
```

```python
# Create new columns to make the dataset easier to analyze

# ANY_INJURY identifies whether a crash resulted in at least one injury
cleaned["ANY_INJURY"] = cleaned["NUMBER OF PERSONS INJURED"] > 0

# ANY_FATALITY identifies whether a crash resulted in at least one death
cleaned["ANY_FATALITY"] = cleaned["NUMBER OF PERSONS KILLED"] > 0

# YEAR extracts the year from the crash date
cleaned["YEAR"] = cleaned["CRASH DATE"].dt.year

# MONTH extracts the month from the crash date
cleaned["MONTH"] = cleaned["CRASH DATE"].dt.month

# DAY_OF_WEEK identifies which day of the week the crash occured
cleaned["DAY_OF_WEEK"] = cleaned["CRASH DATE"].dt.day_name()
```

## Injury Rate by Borough

Which borough has the highest percentage of crashes that result in an injury?

```python
# Calculate injury rate by borough

borough_severity = cleaned[
    cleaned["BOROUGH"] != "Unknown"
].groupby("BOROUGH").agg(
    TOTAL_CRASHES=("COLLISION_ID", "count"),
    INJURY_CRASHES=("ANY_INJURY", "sum")
)

borough_severity["INJURY_RATE"] = (
    borough_severity["INJURY_CRASHES"] /
    borough_severity["TOTAL_CRASHES"] * 100
)

borough_severity = borough_severity.sort_values(
    "INJURY_RATE",
    ascending=False
)

borough_severity
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=borough_severity,
    x=borough_severity.index,
    y="INJURY_RATE"
)

plt.title("Percentage of Crashes Resulting in Injury by Borough")
plt.xlabel("Borough")
plt.ylabel("Crashes Resulting in Injury (%)")

plt.show()
```

### Contributing Factors: Frequency vs Injury Rate
Which common contributing facts are associated with more injuries?

```python
factor_data = cleaned[
    (cleaned["CONTRIBUTING FACTOR VEHICLE 1"].notna()) &
    (cleaned["CONTRIBUTING FACTOR VEHICLE 1"] != "Unspecified")
]

factor_stats = factor_data.groupby(
    "CONTRIBUTING FACTOR VEHICLE 1"
).agg(
    TOTAL_CRASHES=("COLLISION_ID", "count"),
    INJURY_CRASHES=("ANY_INJURY", "sum")
)

factor_stats["INJURY_RATE"] = (
    factor_stats["INJURY_CRASHES"] /
    factor_stats["TOTAL_CRASHES"] * 100
)

# Avoid rare factors with very few observations
factor_stats = factor_stats[
    factor_stats["TOTAL_CRASHES"] >= 1000
]

top_factor_severity = (
    factor_stats
    .sort_values("INJURY_RATE", ascending=False)
    .head(10)
)

top_factor_severity
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_factor_severity,
    x="INJURY_RATE",
    y=top_factor_severity.index
)

plt.title("Contributing Factors With Highest Injury Rates")
plt.xlabel("Crashes Resulting in Injury (%)")
plt.ylabel("Contributing Factor")

plt.show()
```

### Injury Rate by Hour of Day
Now we look at when crashes are more dangerous, not just when more crashes happen.

```python
cleaned["HOUR"] = pd.to_datetime(
    cleaned["CRASH TIME"],
    format="%H:%M",
    errors="coerce"
).dt.hour
```

```python
hour_stats = cleaned.groupby("HOUR").agg(
    TOTAL_CRASHES=("COLLISION_ID", "count"),
    INJURY_CRASHES=("ANY_INJURY", "sum")
)

hour_stats["INJURY_RATE"] = (
    hour_stats["INJURY_CRASHES"] /
    hour_stats["TOTAL_CRASHES"] * 100
)

hour_stats
```

```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=hour_stats,
    x=hour_stats.index,
    y="INJURY_RATE",
    marker="o"
)

plt.title("Crash Injury Rate by Hour of Day")
plt.xlabel("Hour of Day")
plt.ylabel("Crashes Resulting in Injury (%)")

plt.xticks(range(0, 24))

plt.show()
```

### Day of Week x Time of Day Heatmap
What specific days and times experience the highest concentration of collisions?

```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_hour = cleaned.pivot_table(
    index="DAY_OF_WEEK",
    columns="HOUR",
    values="COLLISION_ID",
    aggfunc="count"
)

day_hour = day_hour.reindex(day_order)
```

```python
plt.figure(figsize=(14, 6))

sns.heatmap(
    day_hour,
    cmap="YlOrRd"
)

plt.title("NYC Collisions by Day of Week and Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Day of Week")

plt.show()
```
